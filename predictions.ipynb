```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
## load the trained model

model=load_model('model.h5')

## load encoder and scaler


with open('one_hot_geo.pkl','rb') as file:
    label_encoder_geo=pickle.load(file)

with open('label_encoder_gender.pkl','rb') as file:
    label_rncoder_gender=pickle.load(file)
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
input_data={
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
```

```python
geo_encoded=label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=label_encoder_geo.get_feature_names_out(['Geography']))
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
## encode categorical variables

input_df['Gender']=label_rncoder_gender.transform(input_df['Gender'])
input_df
```

```python
input_df=pd.concat([input_df.drop('Geography',axis=1),geo_encoded_df],axis=1)

input_df
```

```python
## scaling the data
input_scale=scaler.transform(input_df)
input_scale
```

```python
## predic churn

prediction=model.predict(input_scale)
prediction
```

```python
pred_proba=prediction[0][0]

if pred_proba >0.5:
    print('Customer is likely to churn')

else:
    print('customer not likely to churn')
```
